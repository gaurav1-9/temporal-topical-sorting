```python
import pandas as pd
```

## Reading the clustered document

```python
df = pd.read_csv(r"dataset\output\sentences_with_hdbscan_clusters.csv")
print(df.shape)
df.head()
```

```python
print(f"Total columns: {len(df.columns)}")
print(f"Column names: {df.columns.tolist()}")
```

```python
CLUSTER_NUM = 2

filtered_df = df[df['hdb_cluster']==CLUSTER_NUM]
filtered_df = filtered_df[["article_id","sentence","pub_date","hdb_cluster"]]
```

```python
print(f"CLUSTER-{CLUSTER_NUM} Total sentences: {len(filtered_df)}")
filtered_df
```

## Formatting the ***pub_date*** column

```python
import dateparser
```

```python
date_col = "pub_date"

filtered_df['formated_date'] =  filtered_df[date_col].apply(
    lambda x: dateparser.parse(x, settings={
        "TIMEZONE": "UTC", 
        "RETURN_AS_TIMEZONE_AWARE": True
        })
)
filtered_df.head()
```

```python
sorted_df = filtered_df.sort_values(['formated_date'])
sorted_df
```
