```python
import pandas as pd
```

```python
df = pd.read_csv(r"dataset\news_v1.csv")
print(f'''
      Shape: {df.shape}
      Columns: {df.columns.tolist()}
''')
df.head(3)
```

## Data Filtering

```python
target_categories = [
    '5G', 
    'Gadgets'
]
```

```python
filtered_data = df.loc[
    df['category'].isin(target_categories), 
    ['content_id', 'text', 'section', 'published_date', 'category', 'keywords']
].copy()
print(filtered_data.shape)

category_counts = filtered_data['category'].value_counts()
category_counts
```

```python
filtered_data.to_csv(r"dataset\filtering\news_filtered.csv", index=False)
```

## Data Balancing

```python
balanced_data =[]
categories = list(category_counts.keys())
smallest_category = list(min(category_counts.items(), key=lambda x: x[1]))

for i in range(len(categories)):
    if(smallest_category[0] != categories[i]):
        balanced_data.append(
            filtered_data[filtered_data.category == categories[i]]
                .sample(smallest_category[1]+10, random_state=42)
        )
    else:
        balanced_data.append(
            filtered_data[filtered_data.category == categories[i]])
```

```python
tech_data = pd.concat(balanced_data).sample(frac=1, random_state=42).reset_index(drop=True)
tech_data.category.value_counts()
```

```python
tech_data.to_csv(r"dataset\filtering\news_balanced.csv", index=False)
```
