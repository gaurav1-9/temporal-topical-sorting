## Reading the clustered document

```python
df = pd.read_csv(r"dataset\output\sentences_with_hdbscan_clusters.csv")
print(df.shape)
df.head()
```

```python
print(f"Total columns: {len(df.columns)}")
print(f"Column names: {df.columns.tolist()}")
```

```python
clusters = sorted([c for c in df['hdb_cluster'].unique().tolist() if c > -1])
print(clusters)
```

```python
CLUSTER_NUM = clusters[0]

filtered_df = df[df['hdb_cluster']==CLUSTER_NUM].copy()
filtered_df = filtered_df[["article_id","sentence","pub_date","hdb_cluster"]]
```

```python
print(f"CLUSTER-{CLUSTER_NUM} Total sentences: {len(filtered_df)}")
filtered_df.head(10)
```

## Formatting & Sorting the ***pub_date*** column

```python
import dateparser
import os
```

```python
OUTPUT_DIR = r"dataset\output\temporal_meta"
```

```python
for files in os.listdir(OUTPUT_DIR):
    os.remove(os.path.join(OUTPUT_DIR,files))
```

```python
for c in clusters:
    filtered_df = df[df['hdb_cluster']==c].copy()
    filtered_df = filtered_df[["article_id","sentence","pub_date","hdb_cluster"]]
    print(f"CLUSTER-{c} Total sentences: {len(filtered_df)}")
    
    date_col = "pub_date"

    filtered_df['formatted_date'] =  filtered_df[date_col].apply(
        lambda x: dateparser.parse(x, settings={
            "TIMEZONE": "UTC", 
            "RETURN_AS_TIMEZONE_AWARE": True
            })
    )
    
    sorted_df = filtered_df.sort_values(['formatted_date'])
    result_df = sorted_df
    
    output_path = os.path.join(OUTPUT_DIR,f"cluster{c}_temporalSort_metaData.csv")
    result_df.to_csv(output_path,index=False)
```
